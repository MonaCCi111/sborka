## 1. Аппроксимация методом наименьших квадратов

### 1.1 Линейная функция

Для $y=ax+b$ минимизируем сумму квадратов отклонений $L(a,b)=\sum_{i=1}^{n}(ax_i+b-y_i)^2$. Приравняем производные по $a$ и $b$ к нулю.

$$
\begin{cases}
a\sum x_i^2+b\sum x_i=\sum x_i y_i,\\
a\sum x_i+nb=\sum y_i.
\end{cases}
$$

```python
def cramer_line(X, Y):
    n = X.size
    sx = np.round(np.sum(X), 4)
    sy = np.round(np.sum(Y), 4)
    sx2 = np.round(np.sum(np.round(X**2, 4)), 4)
    sxy = np.round(np.sum(np.round(X * Y, 4)), 4)

    delta = np.round(sx2 * n - sx**2, 4)
    delta_a = np.round(sxy * n - sx * sy, 4)
    delta_b = np.round(sx2 * sy - sx * sxy, 4)

    print(f"ΣX = {sx:.4f}, ΣX² = {sx2:.4f}, ΣY = {sy:.4f}, ΣXY = {sxy:.4f}")
    print(f"Δ = {delta:.4f}, Δa = {delta_a:.4f}, Δb = {delta_b:.4f}")

    a = np.round(delta_a / delta, 4)
    b = np.round(delta_b / delta, 4)
    return a, b

a, b = cramer_line(x, y)
print(f"a = {a:.4f}, b = {b:.4f}")
a_lin, b_lin = np.round([a, b], 2)
```

$$

$$

### 1.2 Степенная функция

В $y=\beta x^a$ логарифмируем обе части. Тогда $X=\ln x$, $Y=\ln y$, $Y=aX+b$, где $b=\ln\beta$.

$$
\begin{cases}
18{,}1612a+9{,}2490b=33{,}6696,\\
9{,}2490a+6b=18{,}0285.
\end{cases}
$$

```python
log_x = np.round(np.log(x), 4)
log_y = np.round(np.log(y), 4)

show_table(pd.DataFrame({"x": x, "y": y, "ln x": log_x, "ln y": log_y}))
```

```python
a, b = cramer_line(log_x, log_y)
beta = np.round(np.exp(b), 4)
print(f"a = {a:.4f}, b = {b:.4f}, β = {beta:.4f}")
a_power, beta_power = np.round([a, beta], 2)
```

По формулам Крамера получаем $a\approx1{,}5059$, $b\approx0{,}6835$. Тогда $\beta=e^b\approx1{,}9808$.

$$
\boxed{y_2=1{,}98x^{1{,}51}}
$$

### 1.3 Показательная функция

Для $y=\beta e^{ax}$ получаем $\ln y=ax+\ln\beta$. Подставляем $X=x$ и $Y=\ln y$ в систему для прямой.

$$
\begin{cases}
286a+36b=131{,}7243,\\
36a+6b=18{,}0285.
\end{cases}
$$

```python
a, b = cramer_line(x, log_y)
beta = np.round(np.exp(b), 4)
print(f"a = {a:.4f}, b = {b:.4f}, β = {beta:.4f}")
a_exp, beta_exp = np.round([a, beta], 2)
```

По формулам Крамера получаем $a\approx0{,}3365$, $b\approx0{,}9859$. Тогда $\beta=e^b\approx2{,}6802$.

$$
\boxed{y_3=2{,}68e^{0{,}34x}}
$$

### 1.4 Квадратичная функция

Для $y=ax^2+bx+c$ приравниваем к нулю производные суммы квадратов отклонений по $a$, $b$ и $c$.

$$
\begin{cases}
a\sum x_i^4+b\sum x_i^3+c\sum x_i^2=\sum x_i^2 y_i,\\
a\sum x_i^3+b\sum x_i^2+c\sum x_i=\sum x_i y_i,\\
a\sum x_i^2+b\sum x_i+nc=\sum y_i.
\end{cases}
$$

После подстановки сумм получаем систему.

$$
\begin{cases}
24310a+2556b+286c=15747{,}5,\\
2556a+286b+36c=1700{,}7,\\
286a+36b+6c=199{,}5.
\end{cases}
$$

```python
sx = np.sum(x)
sx2 = np.sum(x**2)
sx3 = np.sum(x**3)
sx4 = np.sum(x**4)

A = np.array([
    [sx4, sx3, sx2],
    [sx3, sx2, sx],
    [sx2, sx, x.size]
])
B = np.array([np.sum(x**2 * y), np.sum(x * y), np.sum(y)])

delta = np.round(np.linalg.det(A), 4)
deltas = []
for i in range(3):
    A_i = A.copy()
    A_i[:, i] = B
    deltas.append(np.round(np.linalg.det(A_i), 4))

delta_a, delta_b, delta_c = deltas
a, b, c = np.round(np.array(deltas) / delta, 4)
print(f"Δ = {delta:.4f}, Δa = {delta_a:.4f}, Δb = {delta_b:.4f}, Δc = {delta_c:.4f}")
print(f"a = {a:.4f}, b = {b:.4f}, c = {c:.4f}")
a_quad, b_quad, c_quad = np.round([a, b, c], 2)
```

a=\frac{81312}{250880}\approx0{,}3241,\quad
b=\frac{829516{,}8}{250880}\approx3{,}3064,\quad
c=\frac{-511212{,}8}{250880}\approx-2{,}0377.
$$

$$
\boxed{y_4=0{,}32x^2+3{,}31x-2{,}04}
$$

## 2. Графики функций и экспериментальные точки

```python
x_plot = np.linspace(x.min(), x.max(), 300)
fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(x_plot, a_lin * x_plot + b_lin, label='Линейная')
ax.plot(x_plot, beta_power * x_plot**a_power, label='Степенная')
ax.plot(x_plot, beta_exp * np.exp(a_exp * x_plot), label='Показательная')
ax.plot(x_plot, a_quad * x_plot**2 + b_quad * x_plot + c_quad, label='Квадратичная')
ax.scatter(x, y, color='black', label='Экспериментальные точки', zorder=3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(alpha=0.25)
ax.legend()
plt.show()
```

## 3. Сравнение результатов

Подставляем исходные $x_i$ в формулы с округлёнными коэффициентами. Для каждой функции считаем суммарную погрешность по исходным $y_i$.

$$
L=\sum_{i=1}^{6}(\hat y_i-y_i)^2.
$$

```python
values = pd.DataFrame({
    'x': x,
    'y': y,
    'Линейная': a_lin * x + b_lin,
    'Степенная': beta_power * x**a_power,
    'Показательная': beta_exp * np.exp(a_exp * x),
    'Квадратичная': a_quad * x**2 + b_quad * x + c_quad
}).round(4)

show_table(values.set_index('x').T)
```

```python
errors = (values.iloc[:, 2:].sub(y, axis=0)**2).sum().round(4)
show_table(errors.to_frame('Суммарная погрешность L'))
```
