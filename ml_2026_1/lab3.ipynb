## Часть 2. Распределение студентов по курсам

```python
group_spec_dict = {
    61 : "МОП",
    62 : "МОП",
    63 : "ТИ",
    64 : "АДИС",
    65: "РС",
    66: "РС",
    67 : "АПР"
}
```

```python
df = df_only.copy()
```

```python
%%capture
assert df.shape[0] == 347, 'В таблице остались повторы или потеряны данные'

assert df.isna().sum().sum() == 0, 'В таблице остались пропуски'
```

```python
df["specialization"] = df["course_3"].map(group_spec_dict).fillna("")
```

```python
df["specialization"].unique()
```

По умолчанию - 1

МОП и ТИ - 2

МИ - 2

```python
df["fall_course_num"] = \
    ((df["is_mi_student"] == 1) | df["specialization"].isin(["МОП", "ТИ"])) \
    .map({True: 2, False: 1}.get)
```

```python
%%capture

col_name = 'fall_course_num'

assert(df[df['id'] == '2662600c2c37e11e62f6ee0b88452f22'][col_name] == 2).all(), 'Тест не пройден'
assert(df[df['id'] == 'd555d2805e1d93d4f023e57dc4c8f403'][col_name] == 1).all(), 'Тест не пройден'
assert(df[df['id'] == '8fe79f84f36e3a5d2d6745621321302c'][col_name] == 1).all(), 'Тест не пройден'
assert(df[df['id'] == 'e4caca755ee0bdd711e18fb8084958b5'][col_name] == 2).all(), 'Тест не пройден'

print("Выполнено")
score += 1.5
```

```python
df.columns
```

```python
df_raspr = df[["id", "percentile", "fall_1", "fall_2", "fall_3", "fall_course_num"]].copy()
df_raspr = df_raspr.sort_values(by="percentile", kind="stable")
df_raspr.head()
```

```python
fall_courses = pd.concat([df_raspr["fall_1"], df_raspr["fall_2"], df_raspr["fall_3"]]).unique()
course_limit = {k: 30 for k in fall_courses}
course_limit["Statistical Learning Theory"] = 60
course_limit["Высокопроизводительные вычисления"] = 60
course_limit["Анализ неструктурированных данных"] = len(df)
```

```python
zachislenie = pd.DataFrame(columns=["id", "course"])
df["is_first_place"] = float("nan")
df["is_last_place"] = float("nan")
```

```python
for c, l in course_limit.items():
    to_add = df_raspr[(df_raspr["fall_1"] == c) |
                      ((df_raspr["fall_course_num"] == 2) & (df_raspr["fall_2"] == c))].head(l)

    if len(to_add) > 0:
        df.loc[df["id"] == to_add.iloc[0]["id"], "is_first_place"] = True
        df.loc[df["id"] == to_add.iloc[-1]["id"], "is_last_place"] = True

    to_add = to_add[["id"]].copy()
    to_add["course"] = c
    course_limit[c] -= len(to_add)
    zachislenie = pd.concat([zachislenie, to_add], ignore_index=True)

zachislenie.head()
```

```python
check_df = pd.read_csv("data/lab03_train_check.csv")
```

```python
%%capture
assert df.loc[df["is_first_place"].notna(), "id"].sort_values().reset_index(drop=True).equals(
    check_df.loc[check_df["is_first_place"].notna(), "id"].sort_values().reset_index(drop=True)
), 'Тест не пройден'
assert df.loc[df["is_last_place"].notna(), "id"].sort_values().reset_index(drop=True).equals(
    check_df.loc[check_df["is_last_place"].notna(), "id"].sort_values().reset_index(drop=True)
), 'Тест не пройден'
print("Выполнено")
score += 2
```

```python
course_count = zachislenie["id"].value_counts()
df_raspr = df_raspr[df_raspr["id"].map(course_count).fillna(0) < df_raspr["fall_course_num"]]
df_raspr.shape
```

```python
df_raspr["next_course"] = df_raspr["fall_2"]
df_raspr.loc[df_raspr["fall_course_num"] == 2, "next_course"] = df_raspr["fall_3"]

for c, l in course_limit.items():
    already = zachislenie.loc[zachislenie["course"] == c, "id"]
    to_add = df_raspr[(df_raspr["next_course"] == c) & ~df_raspr["id"].isin(already)].head(l)
    to_add = to_add[["id"]].copy()
    to_add["course"] = c
    course_limit[c] -= len(to_add)
    zachislenie = pd.concat([zachislenie, to_add], ignore_index=True)
```

```python
rest_1 = df_raspr[(df_raspr["fall_course_num"] == 1) & ~df_raspr["id"].isin(zachislenie["id"])].copy()

for c, l in course_limit.items():
    to_add = rest_1.loc[rest_1["fall_3"] == c, ["id"]].head(l).copy()
    to_add["course"] = c
    course_limit[c] -= len(to_add)
    zachislenie = pd.concat([zachislenie, to_add], ignore_index=True)

zachislenie.shape
```

```python
res_fall = df[["id", "fall_course_num"]].copy()
course1 = zachislenie.groupby("id").head(1).set_index("id")["course"]
course2 = zachislenie[zachislenie.duplicated("id")].set_index("id")["course"]

res_fall["course1"] = res_fall["id"].map(course1).fillna("???")
res_fall["course2"] = res_fall["id"].map(course2).fillna("???")
res_fall.loc[res_fall["fall_course_num"] == 1, "course2"] = "-"
res_fall = res_fall.rename(columns={"id": "ID"})[["ID", "course1", "course2"]]
res_fall.head()
```
